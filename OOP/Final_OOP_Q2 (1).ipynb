### UML 1: Parent Class — `LibraryMember`

```python
# Use a class to define the parent class LibraryMember
class LibraryMember:
    """
    Purpose: Represents a library member using the member ID,
             name, and number of books borrowed. The attributes
             are protected (single underscore) as shown by the
             '#' marker in the class diagram.
    Parameters:
    m_id, n, b.
    Returns:
    None.
    """

    # Use __init__ as a constructor to initialise the instance variables
    def __init__(self, m_id, n, b):
        """
        Purpose: Initialises the values for the member ID, name,
                 and number of books borrowed.
        Parameters:
        m_id, n, b.
        Returns:
        None.
        """
        # Store the member data as protected instance variables
        self._member_id = m_id
        self._name = n
        self._borrowed_count = b

    # Use a method to increase the number of borrowed books
    def borrow_book(self):
        """
        Purpose: Increases the number of books borrowed by the
                 library member.
        Parameters:
        None.
        Returns:
        None.
        """
        # Increase the borrowed book count by one
        self._borrowed_count += 1
```

### UML 2: Child Class — `StudentMember`

```python
# Use inheritance to define StudentMember as a child of LibraryMember
class StudentMember(LibraryMember):
    """
    Purpose: Represents a student library member with courses
             and a borrowing limit. The 'courses' and
             'student_limit' attributes are private (double
             underscore) as shown by the '-' marker in the
             class diagram.
    Parameters:
    m_id, n, b, c.
    Returns:
    None.
    """

    # Use __init__ to initialise inherited and student instance variables
    def __init__(self, m_id, n, b, c):
        """
        Purpose: Initialises the library member values and the
                 courses attended by the student.
        Parameters:
        m_id, n, b, c.
        Returns:
        None.
        """
        # Use super() to initialise variables inherited from LibraryMember
        super().__init__(m_id, n, b)
        # Store the courses as a private instance variable
        self.__courses = c
        # Use len() to set the limit according to the number of courses
        self.__student_limit = len(c)

    # Use a method to control borrowing according to the student limit
    def borrow_book(self):
        """
        Purpose: Allows the student to borrow a book when the
                 borrowing limit has not been reached.
        Parameters:
        None.
        Returns:
        None.
        """
        # Use an if statement to check the student borrowing limit
        if self._borrowed_count < self.__student_limit:
            # Increase the borrowed book count by one
            self._borrowed_count += 1
        else:
            # Display a message when the borrowing limit is exceeded
            print("Borrowing limit exceeded")
```

### UML 3: Child Class — `StaffMember`

```python
# Use inheritance to define StaffMember as a child of LibraryMember
class StaffMember(LibraryMember):
    """
    Purpose: Represents a staff library member with a discipline
             and a borrowing limit. The 'discipline' and
             'staff_limit' attributes are private (double
             underscore) as shown by the '-' marker in the
             class diagram.
    Parameters:
    m_id, n, b, d, sl.
    Returns:
    None.
    """

    # Use __init__ to initialise inherited and staff instance variables
    def __init__(self, m_id, n, b, d, sl):
        """
        Purpose: Initialises the library member values, discipline,
                 and borrowing limit for the staff member.
        Parameters:
        m_id, n, b, d, sl.
        Returns:
        None.
        """
        # Use super() to initialise variables inherited from LibraryMember
        super().__init__(m_id, n, b)
        # Store the staff discipline as a private instance variable
        self.__discipline = d
        # Store the staff borrowing limit as a private instance variable
        self.__staff_limit = sl

    # Use a method to control borrowing according to the staff limit
    def borrow_book(self):
        """
        Purpose: Allows the staff member to borrow a book when
                 the borrowing limit has not been reached.
        Parameters:
        None.
        Returns:
        None.
        """
        # Use an if statement to check the staff borrowing limit
        if self._borrowed_count < self.__staff_limit:
            # Increase the borrowed book count by one
            self._borrowed_count += 1
        else:
            # Display a message when the borrowing limit is exceeded
            print("Borrowing limit exceeded")
```
