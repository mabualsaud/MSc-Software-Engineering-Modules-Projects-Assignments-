**Part (i) and (ii):** Write a new version of the code that has a class called `HotelRoom`. This class should have the data as instance variables and a method that has the same functionality as the function `estimate_cost`. Then write code to instantiate an object of the `HotelRoom` class. A docstring at the start of the class explains its purpose, and clear in-code comments follow the Documentation Requirements.

```python
# Use a class to define HotelRoom
class HotelRoom:
    """
    Purpose: Displays an estimated cost for a hotel room reservation
             based on the number of nights, price per night, and
             breakfast preference.
    """

    # Use __init__ as a constructor to initialise the instance variables
    def __init__(self, name, nights, price, breakfast):
        """
        Purpose: Initializes the values for the guest name, number
                 of nights, price per night, and breakfast preference.
        Parameters:
        name, nights, price, breakfast.
        Returns:
        None.
        """
        # Store the reservation data as instance variables
        self.guest_name = name
        self.number_of_nights = nights
        self.price_per_night = price
        self.breakfast_included = breakfast

    # Use a method to calculate the estimated cost
    def estimate_cost(self):
        """
        Purpose: Calculates the estimated cost by multiplying the
                 number of nights by the price per night, and adds
                 the breakfast cost if selected.
        Parameters:
        None.
        Returns:
        None.
        """
        # Multiply the number of nights by the price per night
        total = self.number_of_nights * self.price_per_night
        # Use an if statement to add the breakfast cost when selected
        if self.breakfast_included:
            total += self.number_of_nights * 12
        # Use print to display the guest name and total cost
        print(self.guest_name, "total cost:", total)


# Instantiate an object from the HotelRoom class
room_reservation = HotelRoom("Olivia", 3, 65, True)

# Call the estimate_cost method using the object
room_reservation.estimate_cost()
```
