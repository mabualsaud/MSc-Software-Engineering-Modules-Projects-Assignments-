# Breadth-first traversal visits the starting vertex first,
# then visits its adjacent unvisited outgoing vertices using a queue.
Start at vertex 5.
Visited: 5
Queue: [5]
Visit vertex 5.
The adjacent unvisited outgoing vertex is 6.
Visited: 5, 6
Queue: [6]
Visit vertex 6.
Vertex 6 has no outgoing unvisited vertices.
Queue: []
The queue is empty, so the traversal is complete.
Breadth-First Traversal:
5 → 6

```python
# ============================================================
# Part (ii): Modify the Graph class in Appendix E to create
# a DirectedWeightGraph, add a dunder method to add two graphs,
# and create the graph shown in the diagram.

# ============================================================
from collections import defaultdict
# Use a class to define a directed weighted graph

class DirectedWeightGraph:
    """
    Purpose: Represents a directed weighted graph containing
             vertices and directed edges with weights.
    """
    # Use __init__ as a constructor to initialise vertices and edges

    def __init__(self):
        """
        Purpose: Initialises an empty set of vertices and an
                 empty collection of directed weighted edges.
        Parameters:
        None.
        Returns:
        None.
        """
        self.vertices = set()
        self.edges = defaultdict(list)
    # Use a method to add a vertex to the graph

    def add_vertex(self, v):
        """
        Purpose: Adds a vertex to the graph.
        Parameters:
        v: The vertex to add.
        Returns:
        None.
        """
        self.vertices.add(v)
    # Use one direction only to create a directed weighted edge

    def add_edge(self, v, u, weight):
        """
        Purpose: Adds a directed weighted edge from one vertex
                 to another vertex.
        Parameters:
        v: The starting vertex.
        u: The destination vertex.
        weight: The weight of the edge.
        Returns:
        None.
        """
        self.edges[v].append((u, weight))
    # Use the __add__ dunder method to combine two graphs

    def __add__(self, other):
        """
        Purpose: Combines two DirectedWeightGraph objects into
                 a new graph.
        Parameters:
        other: The second DirectedWeightGraph object.
        Returns:
        A new DirectedWeightGraph containing the vertices and
        edges from both graphs.
        """
        # Instantiate a new graph for the combined result
        new_graph = DirectedWeightGraph()
        # Use set union to combine vertices from both graphs
        new_graph.vertices = self.vertices | other.vertices
        # Copy the directed weighted edges from the first graph
        for v in self.edges:
            for u, weight in self.edges[v]:
                new_graph.add_edge(v, u, weight)
        # Copy the directed weighted edges from the second graph
        for v in other.edges:
            for u, weight in other.edges[v]:
                new_graph.add_edge(v, u, weight)
        return new_graph

# ============================================================
# Create the DirectedWeightGraph shown in the diagram

# ============================================================
# Instantiate a DirectedWeightGraph object
graph = DirectedWeightGraph()
# Use a loop to add vertices 1 to 6
for vertex in [1, 2, 3, 4, 5, 6]:
    graph.add_vertex(vertex)

# Add the directed weighted edges shown in the graph diagram
graph.add_edge(1, 2, 4)
graph.add_edge(2, 6, 4)
graph.add_edge(3, 1, 2)
graph.add_edge(3, 4, 1)
graph.add_edge(1, 4, 4)
graph.add_edge(4, 1, 3)
graph.add_edge(4, 5, 3)
graph.add_edge(5, 6, 5)
```

```python
# ============================================================
# Example for Validation

# ============================================================
# Display the vertices in the graph
print("Vertices:")
print(graph.vertices)
# Display each directed weighted edge
print("\nDirected weighted edges:")
for v in sorted(graph.edges):
    for u, weight in graph.edges[v]:
        print(v, "->", u, "Weight:", weight)

# Validate the __add__ dunder method using two graphs
graph1 = DirectedWeightGraph()
graph1.add_vertex(1)
graph1.add_vertex(2)
graph1.add_edge(1, 2, 4)
graph2 = DirectedWeightGraph()
graph2.add_vertex(3)
graph2.add_vertex(4)
graph2.add_edge(3, 4, 1)
# Use + to call the __add__ dunder method
combined_graph = graph1 + graph2
print("\nCombined graph vertices:")
print(combined_graph.vertices)
print("\nCombined graph edges:")
for v in sorted(combined_graph.edges):
    for u, weight in combined_graph.edges[v]:
        print(v, "->", u, "Weight:", weight)
```
