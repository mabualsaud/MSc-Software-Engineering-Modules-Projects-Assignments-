```python
# ============================================================
# Part (ii): Write a Schedule class containing a heap instance
# variable using the Heap class in Appendix A. Compare the
# elements based on chronological order and instantiate the
# class using the given list.

# ============================================================
# Use the Heap class from Appendix A and modify the comparisons
# so that time values are compared in chronological order.

class Heap:
    """
    Purpose: Stores time values in a heap and compares them
             according to chronological order.
    """
    # Use __init__ as a constructor to initialise the heap data

    def __init__(self, data=None):
        """
        Purpose: Initialises the data stored in the heap. A
                 mutable default argument is avoided by using
                 None and creating a new list inside the method,
                 so that separate Heap objects do not
                 accidentally share the same list.
        Parameters:
        data.
        Returns:
        None.
        """
        # Use a new empty list when no data is provided, instead
        # of a mutable default argument
        self.data = data if data is not None else []
    # Convert HH:MM into minutes to allow chronological comparison

    def time_value(self, time):
        """
        Purpose: Converts a time value into minutes for
                 chronological comparison.
        Parameters:
        time.
        Returns:
        The time represented as the total number of minutes.
        """
        hours, minutes = time.split(":")
        return int(hours) * 60 + int(minutes)
    # Check whether the heap contains any elements

    def is_empty(self):
        """
        Purpose: Checks whether the heap is empty.
        Parameters:
        None.
        Returns:
        True if the heap is empty, otherwise False.
        """
        return len(self.data) == 0
    # Calculate the parent index of an element

    def parent(self, index):
        """
        Purpose: Calculates the parent index.
        Parameters:
        index.
        Returns:
        The parent index.
        """
        return (index - 1) // 2
    # Calculate the left child index of an element

    def left(self, index):
        """
        Purpose: Calculates the left child index.
        Parameters:
        index.
        Returns:
        The left child index.
        """
        return index * 2 + 1
    # Calculate the right child index of an element

    def right(self, index):
        """
        Purpose: Calculates the right child index.
        Parameters:
        index.
        Returns:
        The right child index.
        """
        return index * 2 + 2
    # Insert a time into the heap and move larger times upward

    def insert(self, value):
        """
        Purpose: Inserts a time value into the heap according
                 to chronological order.
        Parameters:
        value.
        Returns:
        None.
        """
        pos = len(self.data)
        self.data.append(0)
        # Compare the parent and new value using their chronological values
        while (
            pos > 1
            and self.time_value(self.data[self.parent(pos)])
            < self.time_value(value)
        ):
            self.data[pos] = self.data[self.parent(pos)]
            pos = self.parent(pos)
        self.data[pos] = value
    # Extract the largest chronological value from the root

    def extract(self):
        """
        Purpose: Removes and returns the largest chronological
                 time value from the heap.
        Parameters:
        None.
        Returns:
        The largest chronological time value.
        """
        max_value = self.data[0]
        self.data[0] = self.data[len(self.data) - 1]
        self.data.pop()
        # Restore the heap after extracting the root
        self.move_down(0, len(self.data) - 1)
        return max_value
    # Move an element down until the heap condition is restored

    def move_down(self, first, last):
        """
        Purpose: Restores the heap by moving an element downward
                 according to chronological order.
        Parameters:
        first, last.
        Returns:
        None.
        """
        while self.left(first) <= last:
            larger = self.left(first)
            # Compare the left and right children chronologically
            if (
                self.right(first) <= last
                and self.time_value(self.data[self.left(first)])
                < self.time_value(self.data[self.right(first)])
            ):
                larger = self.right(first)
            # Stop when the parent is chronologically larger
            if (
                self.time_value(self.data[first])
                >= self.time_value(self.data[larger])
            ):
                break
            # Swap the parent with the larger child
            temp = self.data[first]
            self.data[first] = self.data[larger]
            self.data[larger] = temp
            first = larger

# Store the list of times given in Question 3
times = ["14:38", "09:12", "18:05", "06:00", "12:10", "17:01"]
# Use a class to define Schedule

class Schedule:
    """
    Purpose: Stores a list of times as a heap.
    """
    # Use __init__ as a constructor to create a heap instance

    def __init__(self, times):
        """
        Purpose: Initialises the schedule using the given list
                 and stores the values in a heap.
        Parameters:
        times.
        Returns:
        None.
        """
        # Instantiate a Heap object as an instance variable
        self.heap = Heap()
        # Use a loop to insert each time into the heap
        for time in times:
            self.heap.insert(time)

# Instantiate a Schedule object using the given list
schedule = Schedule(times)
# Display the heap stored inside the Schedule object
print("Heap:")
print(schedule.heap.data)
# Extract the times from the heap in descending chronological order
print("\nDescending chronological order:")
while not schedule.heap.is_empty():
    print(schedule.heap.extract())
```
