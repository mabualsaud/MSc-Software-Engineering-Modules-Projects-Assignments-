# Identify the depth of each vertex before performing
# the preorder depth-first traversal.
Depth 0: F
Depth 1: D, N
Depth 2: A, G, H
Depth 3: B, J
# Preorder depth-first traversal visits the current vertex first,
# then continues through the left subtree before the right subtree.
Start at vertex F.
Visit F → move to the left vertex D.
Visit D → move to the left vertex A.
Visit A → continue deeper to vertex B.
Visit B → no further vertex, so backtrack to D.
Visit G → the left subtree of F is now complete.
Backtrack to F and move to the right vertex N.
Visit N → continue to H.
Visit H → continue to J.
Visit J → traversal is complete.
Preorder traversal:
F → D → A → B → G → N → H → J

```python
# ============================================================
# Part (ii): Create the tree object using the Node class

# ============================================================

class Node:
    """
    Purpose: Represents a node in a binary tree.
    Parameters:
    val: The value stored in the node.
    l: The left child.
    r: The right child.
    """
    def __init__(self, val, l=None, r=None):
        """
        Purpose: Initialises a node with a value and optional
                 left and right children.
        Parameters:
        val, l, r.
        Returns:
        None.
        """
        self.value = val
        self.left = l
        self.right = r

    def __str__(self):
        """
        Purpose: Returns a string representation of the node
                 and its children.
        Parameters:
        None.
        Returns:
        A string representation of the tree.
        """
        if self.left == None and self.right == None:
            return str(self.value)
        else:
            return str(self.value) + "({})({})".format(
                str(self.left) if self.left != None else "",
                str(self.right) if self.right != None else ""
            )

# Create the leaf nodes first
B = Node("B")
G = Node("G")
J = Node("J")
# Create A with B as its right child
A = Node("A", None, B)
# Create H with J as its right child
H = Node("H", None, J)
# Create D with A as its left child and G as its right child
D = Node("D", A, G)
# Create N with H as its left child
N = Node("N", H, None)
# Create F as the root of the tree
F = Node("F", D, N)
# This tree is NOT a valid binary search tree. In a BST, every
# value in the left subtree of the root must be smaller than the
# root. Node G sits in F's left subtree (as D's right child), but
# G > F alphabetically, which violates the BST property. The right
# subtree of F (N, H, J) does satisfy the BST rule on its own, but
# that alone does not make the whole tree a BST.
print(F)
```

```python
# ============================================================
# Part (iii): Modify the Search_tree class
# ============================================================

# Use the function from Appendix D to check whether the input value is a string.
def is_str(value):
    """
    Purpose: Checks whether a value is a string.
    Parameters:
    value.
    Returns:
    True if the value is a string, otherwise False.
    """
    return isinstance(value, str)

class Search_tree:
    """
    Purpose: Represents a search tree in which uppercase
             characters start in the right subtree of the root
             and lowercase characters start in the left subtree.
    """
    def __init__(self):
        """
        Purpose: Initialises an empty search tree.
        Parameters:
        None.
        Returns:
        None.
        """
        self.root = None

    def insert(self, val, node=None):
        """
        Purpose: Inserts a character into the search tree.
                 Uppercase characters are inserted into the
                 right subtree of the root and lowercase
                 characters into the left subtree.
        Parameters:
        val: The value to insert.
        node: The current node used during insertion.
        Returns:
        The node after insertion.
        Raises:
        ValueError: If the input is not a valid uppercase
                    or lowercase string character.
        """
        # Check that the input is a single string character using Appendix D
        if not is_str(val) or len(val) != 1:
            raise ValueError("The input must be a single string character")

        # Check that the character is uppercase or lowercase
        if not val.isupper() and not val.islower():
            raise ValueError("The input must be an uppercase or lowercase character")

        # Create the root when the tree is empty
        if self.root is None:
            self.root = Node(val)
            return self.root

        # Start the insertion from the root
        if node is None:
            node = self.root

        # At the root, choose the subtree based on character case
        if node == self.root:
            # Insert uppercase characters into the right subtree
            if val.isupper():
                if self.root.right is None:
                    self.root.right = Node(val)
                else:
                    self.insert(val, self.root.right)
            # Insert lowercase characters into the left subtree
            elif val.islower():
                if self.root.left is None:
                    self.root.left = Node(val)
                else:
                    self.insert(val, self.root.left)
        # Once inside the selected subtree, continue as a normal binary search tree
        else:
            # Insert a smaller value into the left subtree
            if val < node.value:
                if node.left is None:
                    node.left = Node(val)
                else:
                    self.insert(val, node.left)
            # Insert a larger value into the right subtree
            elif val > node.value:
                if node.right is None:
                    node.right = Node(val)
                else:
                    self.insert(val, node.right)

        return node

# ============================================================
# Example for Validation
# ============================================================

# Create a Search_tree object to validate the modified insertion rules
tree = Search_tree()

# Insert uppercase and lowercase characters
tree.insert("F")
tree.insert("D")
tree.insert("N")
tree.insert("a")
tree.insert("h")

# Display the tree after insertion
print(tree.root)

# Validate the error handling with an invalid input (integer)
try:
    tree.insert(10)
except ValueError as error:
    print(error)

# Validate the error handling with more than one character
try:
    tree.insert("ABC")
except ValueError as error:
    print(error)
```
